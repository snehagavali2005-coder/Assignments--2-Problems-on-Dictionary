```python
a = {'A': 1, 'B': 2, 'C': 1}
result = {}

for key, value in a.items():
    if value in result:
        result[value].append(key)
    else:
        result[value] = [key]

print(result)
```

```python
D1 = {'a': 10, 'b': 20}
D2 = {'b': 5, 'c': 15}

result = D1.copy()

for key, value in D2.items():
    if key in result:
        result[key] += value
    else:
        result[key] = value

print(result)
```

```python
Marks = {'Amit': 75, 'Neha': 91, 'Ravi': 84}
Threshold = 80

result = {}

for name, marks in Marks.items():
    if marks >= Threshold:
        result[name] = marks

print(result)
```

```python
words = ['apple', 'ant', 'ball', 'bat', 'cat']
result = {}

for word in words:
    first = word[0]

    if first in result:
        result[first].append(word)
    else:
        result[first] = [word]

print(result)
```

```python
D1 = {'a': 1, 'b': 2, 'c': 3}
D2 = {'b': 20, 'c': 3, 'd': 4}

result = set()

for key in D1:
    if key in D2 and D1[key] != D2[key]:
        result.add(key)

print("Different common keys =", result)
```

# **SECTION C -ADVANCED**

```python
words = ['eat', 'tea', 'tan', 'ate', 'nat', 'bat']
result = {}

for word in words:
    key = ''.join(sorted(word))

    if key in result:
        result[key].append(word)
    else:
        result[key] = [word]

print(result)
```

```python
a = "python is easy and python is powerful"
words = a.split()
result = {}

for i, word in enumerate(words):
    if word in result:
        result[word].append(i)
    else:
        result[word] = [i]

print(result)
```

```python
data = {'student': {'name': 'Amit', 'marks': {'Python': 90}}}
result = {}

def flatten(data, prefix=''):
    for key, value in data.items():
        new_key = prefix + key

        if isinstance(value, dict):
            flatten(value, new_key + '.')
        else:
            result[new_key] = value

flatten(data)
print(result)
```

```python
D1 = {'a': 1, 'b': {'x': 10}}
D2 = {'b': {'y': 20}, 'c': 3}

def merge(d1, d2):
    result = d1.copy()

    for key, value in d2.items():
        if key in result and isinstance(result[key], dict) and isinstance(value, dict):
            result[key] = merge(result[key], value)
        else:
            result[key] = value

    return result

result = merge(D1, D2)
print(result)
```

```python
Inventory = {'Pen': 10, 'Book': 5}
Transactions = [('Pen', -3), ('Book', 4), ('Pencil', 6)]

for item, quantity in Transactions:
    if item in Inventory:
        Inventory[item] += quantity
    else:
        Inventory[item] = quantity

print(Inventory)
```

```python
scores = {'A': 95, 'B': 82, 'C': 95, 'D': 70}

sorted_scores = sorted(scores.items(), key=lambda x: x[1], reverse=True)

rank = 0
previous_score = None

for player, score in sorted_scores:
    if score != previous_score:
        rank += 1
        print(f"Rank {rank}:", end=" ")

    print(player, end="")

    same_rank = [
        name for name, marks in sorted_scores
        if marks == score
    ]

    if player == same_rank[-1]:
        print(f" ({score})")

    elif score == previous_score:
        print()

    else:
        print(",", end=" ")

    previous_score = score
```

```python
tasks = {
    'Deploy': ['Test'],
    'Test': ['Build'],
    'Build': ['Code']
}

result = {}

for task, prerequisites in tasks.items():
    for prerequisite in prerequisites:
        if prerequisite in result:
            result[prerequisite].append(task)
        else:
            result[prerequisite] = [task]

print(result)
```

```python
students = {
    'Amit': {'Python': 90, 'ML': 85},
    'Neha': {'Python': 88, 'ML': 94}
}

for name, subjects in students.items():
    subject = max(subjects, key=subjects.get)
    print(name, "->", subject, f"({subjects[subject]})")
```

```python
A = {(0, 0): 2, (1, 1): 3}
B = {(0, 0): 5, (0, 1): 4}

result = A.copy()

for key, value in B.items():
    if key in result:
        result[key] += value
    else:
        result[key] = value

print(result)
```

```python
Capacity = 3
Requests = ['A', 'B', 'C', 'A', 'D']
cache = {}

for item in Requests:
    if item in cache:
        del cache[item]

    cache[item] = True

    if len(cache) > Capacity:
        oldest = next(iter(cache))
        del cache[oldest]

print("Final cache order (oldest to newest) =", list(cache.keys()))
```
